<h3>NumPy (short form of Numerical Python) is an open-source Python library used for
numerical computing. It provides tools for working efficiently with arrays, matrices, and
mathematical functions.</h3>
<h3>NumPy introduces a new data structure called the ndarray (n-dimensional array),
which is much faster and more memory-efficient than Python’s built-in lists for
numerical operations.</h3>

```python
import numpy as np

arr = np.array([1,2,3,"Hello"])
arr
```

```python
#To get size of array
arr.nbytes
```

```python
#Type of array
type(arr)
```

```python
#Shape of array returns in tuple the dimension
arr.shape
```

<h2>Using Bultin Functions lets create ndarray</h2>

```python
#Create arr of (row,col) dimension with 0 values

zerosArr = np.zeros((2,3))
print(zerosArr)

#Creating arr of (row,col) dimension with 1 value

oneArr = np.ones((3,3))
print(oneArr)


#Creating arr of (row,col) dimension with any value

mat = np.full((2,2),5)
print(mat)

#Creating Identity matrix
id_mat = np.eye(3)
print(id_mat)

arr5 = np.arange(1, 20, 2) # Elements in range(1, 20)
print(arr5, arr5.shape)
arr6 = np.linspace(0, 10, 5) # Evenly spaced array
print(arr6, arr6.shape)
print(arr6.ndim)
```
