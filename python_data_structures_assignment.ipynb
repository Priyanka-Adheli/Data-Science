### 2.1 Frequency Counter

Write a function `word_frequency(text)` that takes a string and returns a dictionary mapping each word (lowercase, punctuation-free) to how many times it appears.

**Example:**
```
word_frequency("the cat sat on the mat the cat ran")
# -> {'the': 3, 'cat': 2, 'sat': 1, 'on': 1, 'mat': 1, 'ran': 1}
```

**Hint:** Use `str.split()` to break the text into words, and either a plain `dict` with `.get(word, 0) + 1`, or look up `collections.Counter`.


````python
def word_frequency(text):
    # Your code here
    word_freq = {}

    for word in text.split():
        word_freq[word]=word_freq.get(word,0)+1

    return word_freq

# Test your solution
print(word_frequency("the cat sat on the mat the cat ran"))

# Write at least one more test case of your own below
````

### 2.2 Removing Duplicates While Preserving Order

Sets remove duplicates but don't preserve order. Write a function `dedupe_preserve_order(lst)` that removes duplicates from a list **while keeping the original order** of first occurrence, using a set to track what's already been seen.

**Example:**
```
dedupe_preserve_order([3, 1, 3, 2, 1, 4, 2])
# -> [3, 1, 2, 4]
```

**Hint:** Iterate through the list, keep a `seen = set()`, and only add an item to the result list if it hasn't been seen before.


````python
def dedupe_preserve_order(lst):
    # Your code here
    s1 = set()
    result = []
    for num in lst:
        if num not in s1:
            s1.add(num)
            result.append(num)

    return result


# Test your solution
print(dedupe_preserve_order([3, 1, 3, 2, 1, 4, 2]))

# Write at least one more test case of your own below
````

### Problem 1: Two Sum

**Difficulty:** Medium (classic easy-medium)

Given an array of integers `nums` and an integer `target`, return the **indices** of the two numbers such that they add up to `target`. You may assume each input has exactly one solution, and you may not use the same element twice.

**Example 1:**
```
Input: nums = [2, 7, 11, 15], target = 9
Output: [0, 1]
Explanation: nums[0] + nums[1] == 9
```

**Example 2:**
```
Input: nums = [3, 2, 4], target = 6
Output: [1, 2]
```

**Hint:** A brute-force double loop is O(n²). Instead, walk through the array once, and for each number, check whether `target - number` has **already been seen**, using a `dict` mapping value → index. This gets you to O(n).


````python
def two_sum(nums, target):
    # Your code here
    nums.sort()

    start=0
    end = len(nums)-1

    while start<end:
        sum = nums[start]+nums[end]

        if(sum==target):
            return [start,end]
        
        elif(sum>target):
            end-=1
        else:
            start+=1

# Test your solution
print(two_sum([2, 7, 11, 15], 9))   # expected [0, 1]
print(two_sum([3, 2, 4], 6))        # expected [1, 2]

# Write at least one more test case of your own below
````

```python
def two_sum(nums, target):
    seen = {}  # value -> index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []  # no solution found
```

**Why it works:** We only need one pass because by the time we reach index `i`, every earlier value is already stored in `seen`. Checking `complement in seen` is an O(1) dict lookup, so the whole algorithm runs in O(n) time and O(n) space.

### Problem 2: Group Anagrams

**Difficulty:** Medium

Given an array of strings `strs`, group the anagrams together. You can return the answer in **any order**.

An **anagram** is a word formed by rearranging the letters of another, using all the original letters exactly once.

**Example 1:**
```
Input: strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
Output: [["eat", "tea", "ate"], ["tan", "nat"], ["bat"]]
```

**Example 2:**
```
Input: strs = [""]
Output: [[""]]
```

**Hint:** Two words are anagrams of each other if and only if their **sorted letters** are identical. Use a `dict` where the key is the sorted-letters tuple/string of a word, and the value is a `list` of all original words that match that key.


````python
def group_anagrams(strs):
    # Your code here
    d1 = dict()
    for word in strs:
        w1 = sorted(word)
        d1[w1]+=1

        


# Test your solution
print(group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]))
print(group_anagrams([""]))

# Write at least one more test case of your own below
````

```python
def group_anagrams(strs):
    groups = {}
    for word in strs:
        key = "".join(sorted(word))
        if key not in groups:
            groups[key] = []
        groups[key].append(word)
    return list(groups.values())
```

**Why it works:** Sorting a word's letters produces a canonical form shared by every anagram of that word (e.g. `"eat"`, `"tea"`, and `"ate"` all sort to `"aet"`). Using that canonical form as a dict key lets us group all matching words in a single O(n · k log k) pass, where k is the average word length.

### Problem 3: Longest Consecutive Sequence

**Difficulty:** Medium

Given an unsorted array of integers `nums`, return the length of the longest **consecutive elements sequence**. Your algorithm must run in **O(n)** time (i.e., you cannot simply sort the array).

**Example 1:**
```
Input: nums = [100, 4, 200, 1, 3, 2]
Output: 4
Explanation: The longest consecutive sequence is [1, 2, 3, 4], so the length is 4.
```

**Example 2:**
```
Input: nums = [0, 3, 7, 2, 5, 8, 4, 6, 0, 1]
Output: 9
```

**Hint:** Put all numbers into a `set` for O(1) lookups. Then, only start counting a sequence from a number `n` if `n - 1` is **not** in the set (meaning `n` is the start of a sequence). From each such start, keep checking `n + 1`, `n + 2`, ... while they're in the set, and track the longest run found.


````python
def longest_consecutive(nums):
    # Your code here
    nums.sort()

    count = 1
    maxLen = count

    for i in range(1,len(nums)):
        if nums[i]==nums[i-1]+1:
            count=count+1

        else:
            maxLen=max(maxLen,count)
            count=1

    maxLen=max(maxLen,count)

    return maxLen



# Test your solution
print(longest_consecutive([100, 4, 200, 1, 3, 2]))          # expected 4
print(longest_consecutive([0, 3, 7, 2, 5, 8, 4, 6, 0, 1]))   # expected 9

# Write at least one more test case of your own below
````

```python
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0

    for n in num_set:
        # Only start counting if n is the beginning of a sequence
        if n - 1 not in num_set:
            length = 1
            while n + length in num_set:
                length += 1
            longest = max(longest, length)

    return longest
```

**Why it works:** Checking `n - 1 not in num_set` ensures we only start a scan from the true beginning of each sequence, so even though there's a `while` loop inside a `for` loop, every number is visited at most twice overall (once as a potential start, once while extending a sequence) — giving true O(n) time thanks to O(1) set lookups.
</details>
