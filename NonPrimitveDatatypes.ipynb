# Python List Methods

| Method | Description | Example |
|----------|-------------|---------|
| `append(x)` | Adds an element to the end of the list. | `lst.append(10)` |
| `extend(iterable)` | Adds all elements from an iterable to the end of the list. | `lst.extend([1, 2, 3])` |
| `insert(i, x)` | Inserts an element at a specified position. | `lst.insert(1, 100)` |
| `remove(x)` | Removes the first occurrence of the specified value. | `lst.remove(10)` |
| `pop([i])` | Removes and returns the element at the given index. Defaults to the last element. | `lst.pop()` |
| `clear()` | Removes all elements from the list. | `lst.clear()` |
| `index(x[, start[, end]])` | Returns the index of the first occurrence of a value. | `lst.index(5)` |
| `count(x)` | Returns the number of times a value appears in the list. | `lst.count(5)` |
| `sort(key=None, reverse=False)` | Sorts the list in ascending order by default. | `lst.sort()` |
| `reverse()` | Reverses the order of elements in the list. | `lst.reverse()` |
| `copy()` | Returns a shallow copy of the list. | `new_lst = lst.copy()` |

## Example

```python
numbers = [3, 1, 4, 1, 5]

numbers.append(9)        # [3, 1, 4, 1, 5, 9]
numbers.insert(2, 10)    # [3, 1, 10, 4, 1, 5, 9]
numbers.remove(1)        # Removes first 1
numbers.sort()           # Sorts list
numbers.reverse()        # Reverses list
```

# Python Dictionary Methods

A dictionary (`dict`) stores data as **key-value pairs**.

## Dictionary Methods

| Method | Description | Example |
|----------|-------------|---------|
| `clear()` | Removes all items from the dictionary. | `d.clear()` |
| `copy()` | Returns a shallow copy of the dictionary. | `d.copy()` |
| `fromkeys(iterable, value)` | Creates a new dictionary with specified keys and value. | `dict.fromkeys(['a','b'], 0)` |
| `get(key, default)` | Returns value for key; default if key not found. | `d.get('name')` |
| `items()` | Returns a view of key-value pairs. | `d.items()` |
| `keys()` | Returns a view of keys. | `d.keys()` |
| `values()` | Returns a view of values. | `d.values()` |
| `pop(key[, default])` | Removes and returns value of specified key. | `d.pop('age')` |
| `popitem()` | Removes and returns the last inserted key-value pair. | `d.popitem()` |
| `setdefault(key[, default])` | Returns value if key exists, else inserts key with default value. | `d.setdefault('city', 'Mumbai')` |
| `update(other)` | Updates dictionary with another dictionary or iterable. | `d.update({'age': 22})` |

## Examples

```python
d = {
    "name": "Priyanka",
    "age": 22
}

print(d.get("name"))          # Priyanka
print(d.keys())               # dict_keys(['name', 'age'])
print(d.values())             # dict_values(['Priyanka', 22])

d.update({"city": "Mumbai"})
print(d)

d.pop("age")
print(d)

| Function       | Description                             | Example                   |
| -------------- | --------------------------------------- | ------------------------- |
| `len()`        | Number of key-value pairs.              | `len(d)`                  |
| `max()`        | Maximum key.                            | `max(d)`                  |
| `min()`        | Minimum key.                            | `min(d)`                  |
| `sorted()`     | Returns sorted list of keys.            | `sorted(d)`               |
| `reversed()`   | Reverse iteration over keys.            | `reversed(d)`             |
| `any()`        | True if any key is truthy.              | `any(d)`                  |
| `all()`        | True if all keys are truthy.            | `all(d)`                  |
| `enumerate()`  | Enumerates keys.                        | `enumerate(d)`            |
| `dict()`       | Creates a dictionary.                   | `dict(a=1, b=2)`          |
| `zip()`        | Combine iterables into key-value pairs. | `dict(zip(keys, values))` |
| `isinstance()` | Checks if object is a dictionary.       | `isinstance(d, dict)`     |

# Python Set Methods

A **set** is an unordered collection of unique elements.

## Set Methods

| Method | Description | Example |
|----------|-------------|---------|
| `add(x)` | Adds an element to the set. | `s.add(10)` |
| `clear()` | Removes all elements from the set. | `s.clear()` |
| `copy()` | Returns a shallow copy of the set. | `s.copy()` |
| `difference(*others)` | Returns elements present in the set but not in others. | `s.difference(t)` |
| `difference_update(*others)` | Removes elements found in other sets. | `s.difference_update(t)` |
| `discard(x)` | Removes element if present; no error if absent. | `s.discard(10)` |
| `intersection(*others)` | Returns common elements. | `s.intersection(t)` |
| `intersection_update(*others)` | Keeps only common elements. | `s.intersection_update(t)` |
| `isdisjoint(other)` | Returns `True` if sets have no common elements. | `s.isdisjoint(t)` |
| `issubset(other)` | Returns `True` if set is a subset of another. | `s.issubset(t)` |
| `issuperset(other)` | Returns `True` if set is a superset of another. | `s.issuperset(t)` |
| `pop()` | Removes and returns an arbitrary element. | `s.pop()` |
| `remove(x)` | Removes element; raises `KeyError` if absent. | `s.remove(10)` |
| `symmetric_difference(other)` | Returns elements present in exactly one set. | `s.symmetric_difference(t)` |
| `symmetric_difference_update(other)` | Updates set with symmetric difference. | `s.symmetric_difference_update(t)` |
| `union(*others)` | Returns union of sets. | `s.union(t)` |
| `update(*others)` | Adds all elements from others. | `s.update(t)` |

## Examples

```python
s = {1, 2, 3}
t = {3, 4, 5}

print(s.union(t))                # {1, 2, 3, 4, 5}
print(s.intersection(t))         # {3}
print(s.difference(t))           # {1, 2}
print(s.symmetric_difference(t)) # {1, 2, 4, 5}

s.add(10)
s.remove(2)

| Method                   | Time Complexity       |
| ------------------------ | --------------------- |
| `add()`                  | O(1) Average          |
| `remove()`               | O(1) Average          |
| `discard()`              | O(1) Average          |
| `pop()`                  | O(1) Average          |
| `union()`                | O(len(s)+len(t))      |
| `intersection()`         | O(min(len(s),len(t))) |
| `difference()`           | O(len(s))             |
| `symmetric_difference()` | O(len(s)+len(t))      |
| `issubset()`             | O(len(s))             |
| `issuperset()`           | O(len(t))             |
| `isdisjoint()`           | O(min(len(s),len(t))) |
