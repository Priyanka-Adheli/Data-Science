```python
color = input("Enter your fav color")

match color:
    case "red":
        print("Stop")
    case "yellow":
        print("Look")
    case "green":
        print("Go")
    case _:
        print("wrong color")
```

<h2>Loops - While and For loop</h2>

```python
#while - repeate until condition is true

i=0

while i<=5:
    print(i)
    i=i+1
    
```

```python
#for loop - to iterate over the sequence like list,tuple,string,range

for i in range(0,6):
    print(i)
```

```python
#in and not in to check memebership operators

word="helloworld"

# for i in word:
#     print(i)


# for i in range(0,len(word)):
#     print(word[i])


for i in range(0,len(word)):
    if word[i] not in 'aeiou':
        print(word[i])


vowels_count = 0;
consonants_count = 0;

for i in range(0,len(word)):
    if word[i] in "aeiou":
        vowels_count=vowels_count+1
    else:
        consonants_count=consonants_count+1


print(f"In word {word} the vowels count is {vowels_count} and consonants count is {consonants_count}")
```

<h2>Functions - def keyword </h2>

```python
def fact(n):
    if n<=1:
        return n
    return n*fact(n-1)



print(fact(5))
```
