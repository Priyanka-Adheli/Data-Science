| Mode | Meaning | Usage Example |
| --- | --- | --- |
| ``"r"`` | Read (default). Opens file for reading. Error if file doesn’t exist. | ``open("file.txt", ``"r")`` |
| ``"w"`` | Write. Creates new file or overwrites existing file. | ``open("file.txt", ``"w")`` |
| ``"a"`` | Append. Opens file for writing, adds new content at the end. | ``open("file.txt", ``"a")`` |
| ``"x"`` | Exclusive creation. Creates new file, error if file exists. | ``open("file.txt", ``"x")`` |
| ``"b"`` | Binary mode. Used with other modes for non‑text files (images, audio). | ``open("file.png", ``"rb")`` |
| ``"t"`` | Text mode (default). Used with other modes for text files. | ``open("file.txt", ``"rt")`` |
| ``"+"`` | Read and write. Combines with other modes. | ``open("file.txt", ``"r+")`` |

<h2>Introduction To JSON</h2>

<h4>JSON (JavaScript Object Notation) is a lightweight data format used to exchange data
between programs, APIs, websites, etc. JSON format is very similar to Python
dictionaries.</h4>

```python
import json

person = {
    "name":"Priyanka",
    "age":21,
    "role":"SDE"
}

#To convert py dict into json use dumps(only 1 dict obj)

json_obj = json.dumps(person)

json_obj
```

```python
#To convert json to py dict use load

person_info = json.loads(json_obj)

person_info
```

```python
type(person_info)
```

<h4>Handling Files with Json data</h4>

```python
f = open("person_info.txt","w")
```

```python
json.dump(person,f,indent=4)
```

```python
f1 = open("person_info.txt","r")

data = json.load(f1)

data
```
