<h3>The self keyword refers to instance (object) of class that is calling method</h3>

```python
class Products:
    # def __init__(self):
    #     print("Default constructor called")

    #If two or more  constructors specified the latest one is called
    # def __init__(self,name):
    #     print("Parameterized Constructor called")    


    #constructor with params
    def __init__(self,name,price,quantity):
        self.name = name
        self.price = price
        self.quantity=quantity

    #Class attribute - common to all the objects of class
    shipping_route = "ABC Road Line"

    #Class attribute can be modifed using @classmethod decorator
    #@classmethod takes cls as first default agrument

    @classmethod
    def changeShippingRoute(cls,new_route):
        cls.shipping_route=new_route

    #@staticmethod decorator makes any function specific to class and it doesnt have self,cls as default args
    @staticmethod
    def classInfo():
        print("This is an Product class")
```

```python
p1 = Products("Asymertic Top",120,1)
p2 = Products("cetaphil",120,2)
```

```python
print(p1.name)
print(p1.price)
print(p1.quantity)
```

```python
p1.changeShippingRoute("PR Road")
```

<h2>Public,Private and Protected Access Specifier</h2>

```python
class Employee:
    def __init__(self,name,salary,department):
        self.name = name # name is public
        self.__salary = salary #Salary is private
        self._department = department #department is protected


e1 = Employee("Raj",900000,"SDE")
```

```python
print(e1.name)
```

```python
# print(e1.__salary) #Error
print(e1._Employee__salary)
```

How it works<br>
By convention, a single underscore (_attr) means “protected”:<br>

It signals to other developers that this attribute is internal and shouldn’t be touched outside the class hierarchy.<br>

But Python doesn’t enforce strict access control — it’s just a naming convention.<br>

Subclasses (child classes) can freely access and override protected members.<br>

This access continues down the inheritance chain — meaning all levels of subclasses can access protected members.<br>

```python
print(e1._department)
```
