How it works<br>
By convention, a single underscore (_attr) means “protected”:<br>

It signals to other developers that this attribute is internal and shouldn’t be touched outside the class hierarchy.<br>

But Python doesn’t enforce strict access control — it’s just a naming convention.<br>

Subclasses (child classes) can freely access and override protected members.<br>

This access continues down the inheritance chain — meaning all levels of subclasses can access protected members.<br>

<h4>Inheritance - Single Inheritence</h4>

````python
class Parent:
    info = "This is parent class"

class Child(Parent):
    def __init__(self,info):
        self.info = info

    def parent_info(self):
        print(super().info)

c1 = Child("This is object 1")

c1.info

c1.parent_info()
        
````

<h4>Multi-level Inheritence </h4>

````python
class GrandParent:
    info="This is grandParent class"

class Parent(GrandParent):
    info = "This is parent Info"

class Child(Parent):
    def __init__(self):
        print("Child class object created")

    def parent_info(self):
        print(super().info)

    def grandParent_info(self):
        print(super(Parent, self).info)


obj = Child()
````

````python
obj.parent_info()
````

````python
obj.grandParent_info()
````

<h4>Multiple Inheritence</h4>

````python
class Person:
    def __init__(self,name):
        self.name = name

class Worker(Person):
    def __init__(self,name,job_title):
        super().__init__(name)
        self.job_title = job_title

    def info(self):
        print(f"My name is {self.name} and i am an worker as {self.job_title}")


class Employee(Person):
    def __init__(self,name,department):
        super().__init__(name)
        self.department = department

    def info(self):
        print(f"My name is {self.name} and I am an Employee working under department {self.department}")

````

````python
# Usage
w1 = Worker("Priyanka", "Data Scientist")
w1.info()

e1 = Employee("Priyanka", "IT")
e1.info()
````

### 🔑 `super()` — What It Can and Cannot Access
- ✅ **Can access:**
  - Parent **methods** (instance methods, class methods, static methods).
    - Example: `super().__init__(name)` calls the parent constructor.
    - Example: `super().info()` calls a parent’s `info()` method.
  - Parent **class-level attributes** (variables defined directly in the class, not inside `__init__`).
    - Example:  
      ```python
      class Person:
          species = "Human"
      class Worker(Person):
          def show_species(self):
              print(super().species)  # Works
      ```

- ❌ **Cannot access:**
  - Parent **instance attributes** (like `self.name`) because they are stored in the object, not the class.
    - `super().name` → ❌ AttributeError

---

### 🧑‍💻 How `self.name` is Accessed in Child Classes
- `self.name` is created inside `Person.__init__`.
- When a child class (`Worker` or `Employee`) calls `super().__init__(name)`, it runs the parent constructor, which sets `self.name` on the object.
- Since the child inherits from `Person`, the same object (`self`) carries that attribute forward.
- That’s why you can directly use `self.name` inside child methods.

---

### ⚡ Quick Example

```python
class Person:
    def __init__(self, name):
        self.name = name

class Worker(Person):
    def __init__(self, name, job_title):
        super().__init__(name)   # initializes self.name
        self.job_title = job_title

    def info(self):
        print(f"My name is {self.name} and I am a worker as {self.job_title}")
```

✔️ `self.name` works because it was set by the parent constructor.  
✔️ `super().__init__(name)` ensures the parent’s initialization runs.  
❌ `super().name` fails because `name` is not a class attribute.

---

👉 Think of it like this:  
- `self` → the **object’s data** (instance attributes).  
- `super()` → the **parent’s behavior** (methods and class attributes).

<h5>Abstract Method</h5>
It is a method declared but not implemented (Children must override abstract methods).<br>

```python

@abstractmethod
def method_name(self):

````python
from abc import ABC,abstractmethod

class Shape(ABC):
    @abstractmethod
    def cal_area(self):
        pass


class Rectangle(Shape):

    def cal_area(self):
        print("Area of Rectangle")

class Square(Shape):
    def cal_area(self):
        print("Area of Square")
````

````python
s1 = Square()
s1.cal_area()
````

# Polymorphism

Polymorphism is the ability of a single function, operator, or object to behave differently based on the context.  
*(“poly” = many, “morph” = forms)*

---

## Key Concepts
- **Same method name** → works differently for different objects  
- **Same operator** → behaves differently depending on operand types  

---

## Example: Operator Overloading
```python
print(1 + 2)       # adds 2 numbers
print("1" + "2")   # concatenates 2 strings
```
Here, the same `+` operator is used for two different operations.

---

## Types of Polymorphism

### 1. Function Overriding (Method Overriding)
- A child class provides its own version of a method that already exists in the parent class.  
- Both methods must have the **same name**.  
- This is a type of **Runtime Polymorphism (dynamic binding)**.  
- The child method takes precedence over the parent method.  

---

### 2. Duck Typing
- Works on the principle:  
  *“If it looks like a duck and quacks like a duck, it must be a duck.”*  
- Focuses on behavior rather than the actual type of the object.  

<h4>Method Overriding</h4>

````python
class Animal:
    def info(self):
        print("I am an Animal Class")

class Cat(Animal):
    def info(self):
        print("I am Cat")

a1 = Animal()
c1 = Cat()
````

````python
a1.info()
````

````python
c1.info()
````

<h4>Duck Typing</h4>

````python
class Dog:
    def speak(self):
        print("Bark !!!")

class Cat:
    def speak(self):
        print("Meoww !!!")

class Bird:
    def speak(self):
        print("Chiww Chiwww !!!")


def make_it_speak(entity):
    entity.speak()


d1 = Dog()
c1 = Cat()
b1 = Bird()

for entity in [d1,c1,b1]:
    make_it_speak(entity)
````
